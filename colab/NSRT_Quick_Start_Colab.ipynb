# Star-Sensor Vaneless Baffle — NSRT Quick Start

This notebook is the browser-based entry point for engineers and researchers who want to inspect or reuse the NSRT method without setting up Python locally.

Repository: https://github.com/mechvivek-explore/star-sensor-vaneless-baffle

## Quick route
1. Install the dependencies.
2. Load `data/full_nsrt_results.csv`.
3. Check ray conservation.
4. Plot pupil-reaching ray count versus off-axis angle.
5. Open `code/nsrt_3d.py` for reuse and development.

The archived calculations are ideal-specular geometrical NSRT results, not PST/NPST or detector-radiometric predictions.
